```python
# Check the month column to calculate Binominal Distribution
print(df['month'].head(10))
print("\nData type:")
print(df['month'].dtype)
print("\nAvailable months:")
print(df['month'].unique()[:15])
```

```python
print(df['month'].unique()[-15:])
```

```python
# Select one month from the dataset because we need it to calculate binominal distribution
monthly_data = df[df['month'] == 202406]
display(monthly_data.head())
# Count the observations
print("Number of observations:", len(monthly_data))
```

```python
print(monthly_data['cardholder_location'].nunique())
```

```python
# Calculate the median of market spend index
median_spend = monthly_data['market_spend_index'].median()
print("Median:", median_spend)
# Calculate the probability of success
p = (monthly_data['market_spend_index'] > median_spend).mean()
print("Probability of success:", p)
```

```python
from scipy.stats import binom
```

```python
# Calculate Binomial probability
n = 20
k = 8
probability = binom.pmf(k, n, p)
print("Probability of exactly 8 successes:", probability)
```

```python
# Plot the binomial distribution
k_values = np.arange(0, n + 1)
probabilities = binom.pmf(k_values, n, p)
sns.barplot(x=k_values, y=probabilities)
plt.title('Binomial Distribution> Market Spend Index')
plt.xlabel('Number of Successes')
plt.ylabel('Probability')
plt.show()
```

```python
# Binomial Distribution with a larger sample
n = 100
k_values = np.arange(0, n + 1)
probabilities = binom.pmf(k_values, n, p)
sns.barplot(x=k_values, y=probabilities)
plt.title('Binomial Distribution - Large Sample (n=100)')
plt.xlabel('Number of successes')
plt.ylabel('Probability')
plt.show()
```

```python
from scipy.stats import poisson
```

```python
# What is the probability of observing a certain number of geographical locations with a market spend index above the overall median during one month?.
#at this point I have to do a liitle research by myself to understand how to apply Poisson Distribution properly and I find this code which can help us to group the records by month.
# Calculate the overall median
median_spend = df['market_spend_index'].median()
# Count locations above the median for each month
monthly_events = df[df['market_spend_index'] > median_spend].groupby('month').size()
print(monthly_events)
```

```python
# Calculate the average number of events
lambda_value = monthly_events.mean()
print("Lambda:", lambda_value)
```

```python
#parameters
k = np.arange(4400, 5050)
# Poisson probability mass function
poisson_pmf = poisson.pmf(k, lambda_value)
# Plot the Poisson distribution
plt.bar(k, poisson_pmf, color='green',edgecolor='black')
plt.title('Poisson Distribution - Market Spend Index')
plt.xlabel('Number of Events')
plt.ylabel('Probability')
plt.show()
```

```python
# Calculate mean and variance of monthly events
print("Mean:", monthly_events.mean())
print("Variance:", monthly_events.var())
```

```python
# Poisson Distribution with a larger Lambda
lambda_large = 6000
k = np.arange(5500, 6500)
poisson_pmf = poisson.pmf(k, lambda_large)
plt.bar(k, poisson_pmf, color='green',edgecolor='black')
plt.title('Poisson Distribution - Larger Lambda (6000)')
plt.xlabel('Number of Events')
plt.ylabel('Probability')
plt.show()
```
